# Evaluation Report

Load metrics JSON artifacts and summarize key results across models and blends.


```python
import os, json, glob
import pandas as pd

root = os.path.dirname(os.path.dirname(os.path.abspath(__file__)))
artifacts = os.path.join(root, 'artifacts')
files = sorted(glob.glob(os.path.join(artifacts, 'metrics*.json'))) or [os.path.join(artifacts, 'metrics.json')]

records = []
for fp in files:
    try:
        with open(fp, 'r', encoding='utf-8') as f:
            data = json.load(f)
        for model, metrics in data.items():
            row = {'artifact': os.path.basename(fp), 'model': model}
            row.update(metrics)
            records.append(row)
    except Exception as e:
        print('Failed to load', fp, e)

if records:
    df = pd.DataFrame(records)
    display(df.sort_values(by=['ndcg@20'], ascending=False).head(20))
else:
    print('No metrics artifacts found yet. Run src/recsys/pipeline_demo.py.')
```
