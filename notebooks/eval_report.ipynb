# Evaluation Report

Load metrics JSON artifacts and summarize key results across models and blends.


```python
import os, json, glob
import pandas as pd

root = os.path.dirname(os.path.dirname(os.path.abspath(__file__)))
exp_root = os.path.join(root, 'experiments')
exp_dirs = sorted([d for d in glob.glob(os.path.join(exp_root, '*')) if os.path.isdir(d) and os.path.basename(d) != 'latest'])

records = []
for d in exp_dirs:
    fp = os.path.join(d, 'metrics.json')
    try:
        with open(fp, 'r', encoding='utf-8') as f:
            data = json.load(f)
        for model, metrics in data.items():
            row = {'experiment': os.path.basename(d), 'model': model}
            row.update(metrics)
            records.append(row)
    except Exception as e:
        print('Failed to load', fp, e)

if records:
    df = pd.DataFrame(records)
    display(df.sort_values(by=['ndcg@20', 'recall@20'], ascending=[False, False]).head(50))
else:
    print('No experiments found yet. Run src/recsys/pipeline_demo.py to create one.')
```
